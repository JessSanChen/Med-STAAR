```python
import numpy as np
import pandas as pd
```

```python
avg_volumes_df = pd.read_csv('facility_volume.csv')
```

```python
display(avg_volumes_df)
```

```python
n_days = 365
start_date = pd.Timestamp("2025-01-01")
```

```python
col_map = {
    "facility_name": "facilities",
    "Volume MD1": "avg_md1",
    "Volume MD2": "avg_md2",
    "Volume PM": "avg_pm",
}

avg_volumes_df = avg_volumes_df.rename(columns=col_map)

display(avg_volumes_df)
```

```python
for c in ["avg_md1", "avg_md2", "avg_pm"]:
    avg_volumes_df[c] = pd.to_numeric(avg_volumes_df[c].replace("NC", np.nan), errors="coerce")
    
display(avg_volumes_df)
```

```python
# avg_volumes_df = avg_volumes_df.set_index('facility')
# avg_volumes_df.index.is_unique

print(avg_volumes_df)
avg_volumes_df.to_csv('cleaned_average_volume.csv')
```

```python
cleaned_avg_volumes_df = pd.read_csv('cleaned_average_volume.csv')
display(cleaned_avg_volumes_df)
```

```python
# skip facilities/shifts with NaN

keep = (cleaned_avg_volumes_df[['avg_md1','avg_md2','avg_pm']].fillna(0) > 0).any(axis=1)
cleaned_avg_volumes_df = cleaned_avg_volumes_df[keep]
display(cleaned_avg_volumes_df)
```

```python
synthetic_data = []
rng = np.random.default_rng(123)


for row in cleaned_avg_volumes_df.itertuples(index=False):
    facility = row.facility
    avg_md1  = float(row.avg_md1 or 0.0)
    avg_md2  = float(row.avg_md2 or 0.0)
    avg_pm   = float(row.avg_pm  or 0.0)

    for day in range(n_days):
        date = start_date + pd.Timedelta(days=day)
        seasonal_factor = 1 + 0.2 * np.sin(2 * np.pi * day / 365)
        
#         print(pd.isnull(avg_md1))
        
        if pd.isnull(avg_md1):
            md1 = float("nan")
        else:
            md1 = rng.poisson(lam=max(avg_md1 * seasonal_factor, 0))
            
        if pd.isnull(avg_md2):
            md2 = float("nan")
        else:
            md2 = rng.poisson(lam=max(avg_md2 * seasonal_factor, 0))
            
        if pd.isnull(avg_pm):
            pm = float("nan")
        else:
            pm = rng.poisson(lam=max(avg_pm * seasonal_factor, 0))

#         md1 = rng.poisson(lam=max(avg_md1 * seasonal_factor, 0))
#         md2 = rng.poisson(lam=max(avg_md2 * seasonal_factor, 0))
#         pm  = rng.poisson(lam=max(avg_pm  * seasonal_factor, 0))

        synthetic_data.append([facility, date, md1, md2, pm])

synthetic_df = pd.DataFrame(synthetic_data, columns=["facility", "date", "md1", "md2", "pm"])
display(synthetic_df)
```

```python
synthetic_df.to_csv("synthetic_data.csv")
```

```python
# Ensure proper dtypes
synthetic_df['date'] = pd.to_datetime(synthetic_df['date'])
# Keep only the columns we need and melt to long format for convenience
value_cols = ['md1', 'md2', 'pm']
df_long = synthetic_df.melt(id_vars=['facility', 'date'], value_vars=value_cols,
                  var_name='shift', value_name='volume')
# Sort for sanity
df_long = df_long.sort_values(['facility', 'shift', 'date']).reset_index(drop=True)
```

```python
display(df_long)
```

```python
import os
import matplotlib.pyplot as plt


os.makedirs("figs", exist_ok=True)

for facility, g in df_long.groupby('facility'):
    plt.figure(figsize=(12, 5))
    for shift, gs in g.groupby('shift'):
        plt.plot(gs['date'], gs['volume'], label=shift, linewidth=2)
    plt.title(f"Daily Consult Volume — {facility}")
    plt.xlabel("Date")
    plt.ylabel("Consults")
    plt.legend(title="Shift")
    plt.tight_layout()
    out_path = os.path.join("figs", f"{facility}_consult_volume.png")
    plt.savefig(out_path, dpi=150)
    plt.close()

print("Saved facility line charts to ./figs/")
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
```

```python
def fit_forecast_series(series: pd.Series, horizon: int = 31):
    """
    series: pd.Series indexed by date (daily), values = volume (float)
    Returns: DataFrame with forecast mean and 95% conf. intervals
    """
    # Ensure daily frequency and fill missing days with zeros (common for count-like data)
    s = series.asfreq('D').fillna(0.0)

    # Clip negative input just in case
    s = s.clip(lower=0.0)

    # Minimal SARIMAX with weekly seasonality:
    # - Non-seasonal (1,0,0)
    # - Seasonal (0,1,1, 7) to capture weekly pattern and seasonal differencing
    # If series is extremely short/flat, fall back gracefully.
    try:
        mod = SARIMAX(
            s,
            order=(1, 0, 0),
            seasonal_order=(0, 1, 1, 7),
            enforce_stationarity=False,
            enforce_invertibility=False
        )
        res = mod.fit(disp=False)
    except Exception as e:
        # Fallback: naive (last observed) if model fails
        last_val = s.iloc[-1] if len(s) else 0.0
        future_index = pd.date_range(s.index.max() + pd.Timedelta(days=1), periods=horizon, freq='D')
        return pd.DataFrame({
            'date': future_index,
            'yhat': last_val,
            'yhat_lower': np.maximum(0.0, last_val - 0.25 * abs(last_val)),
            'yhat_upper': last_val + 0.25 * abs(last_val)
        })

    # Forecast
    fc = res.get_forecast(steps=horizon)
    mean = fc.predicted_mean.clip(lower=0.0)
    ci = fc.conf_int(alpha=0.05)  # 95% CI
    # Some statsmodels versions name columns like 'lower volume'/'upper volume' or 'lower y'/'upper y'
    lower_col = [c for c in ci.columns if 'lower' in c.lower()][0]
    upper_col = [c for c in ci.columns if 'upper' in c.lower()][0]

    out = pd.DataFrame({
        'date': mean.index,
        'yhat': mean.values,
        'yhat_lower': ci[lower_col].clip(lower=0.0).values,
        'yhat_upper': ci[upper_col].clip(lower=0.0).values
    })
    return out

horizon = 31
all_forecasts = []

# Identify last historical date globally to start all horizons consistently
last_hist_date = df_long['date'].max()

for (facility, shift), g in df_long.groupby(['facility', 'shift']):
    # Build a clean time series indexed by date
    ts = g.set_index('date')['volume'].astype(float).sort_index()
    # Ensure we only forecast after the max historical date (in case of gaps)
    forecast_df = fit_forecast_series(ts, horizon=horizon)
    forecast_df.insert(0, 'facility', facility)
    forecast_df.insert(1, 'shift', shift)
    all_forecasts.append(forecast_df)

forecasts = pd.concat(all_forecasts, ignore_index=True)

# ----------------------------
# 4) Save forecasts
# ----------------------------
out_csv = "forecasts_31d.csv"
forecasts.to_csv(out_csv, index=False)

print(f"Saved 31-day forecasts to ./{out_csv}")

# ----------------------------
# 5) (Optional) Quick plot per facility of forecasts vs last 30d history
# ----------------------------
os.makedirs("figs/forecasts", exist_ok=True)

lookback_days = 30
for facility, gfac in df_long.groupby('facility'):
    # One panel per shift to keep it clean
    for shift, gshift in gfac.groupby('shift'):
        hist = gshift.set_index('date')['volume'].astype(float).sort_index()
        hist_tail = hist.loc[hist.index.max() - pd.Timedelta(days=lookback_days-1):]

        fcs = forecasts[(forecasts['facility'] == facility) & (forecasts['shift'] == shift)].copy()
        fcs = fcs.set_index('date').sort_index()

        plt.figure(figsize=(12, 5))
        # history
        if not hist_tail.empty:
            plt.plot(hist_tail.index, hist_tail.values, label='History (last 30d)', linewidth=2)
        # forecast mean
        plt.plot(fcs.index, fcs['yhat'].values, label='Forecast (mean)', linewidth=2)
        # confidence band
        plt.fill_between(fcs.index, fcs['yhat_lower'].values, fcs['yhat_upper'].values,
                         alpha=0.2, label='95% interval')

        plt.title(f"{facility} — {shift}: 30d History + 31d Forecast")
        plt.xlabel("Date")
        plt.ylabel("Consults")
        plt.legend()
        plt.tight_layout()

        out_path = os.path.join("figs", "forecasts", f"{facility}_{shift}_forecast.png")
        plt.savefig(out_path, dpi=150)
        plt.close()

print("Saved forecast preview charts to ./figs/forecasts/")
```
